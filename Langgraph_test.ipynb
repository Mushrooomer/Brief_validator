```python
import os, getpass

def _set_env(var: str):
    if not os.environ.get(var):
        os.environ[var] = getpass.getpass(f"{var}: ")

_set_env("OPENAI_API_KEY")
```

```python
from langchain_openai import ChatOpenAI
gpt4o_chat = ChatOpenAI(model="gpt-4o", temperature=0)
gpt35_chat = ChatOpenAI(model="gpt-3.5-turbo-0125", temperature=0)
```

```python
email = """Dobrý den Karle,
děkujeme za váš e-mail ohledně připravované marketingové kampaně. Bohužel momentálně nemáme k dispozici všechny informace, které jste požadovali, a některé údaje stále interně řešíme. Přesto zde přikládáme několik bodů, které víme, i když nejsou zatím zcela kompletní:

Cíl kampaně
Posílení obecného povědomí o naší značce.
… (Detaily cílů se ještě finalizují, nemáme jasně definované metriky úspěšnosti.)

Cílové publikum
Obecně se zaměřujeme na dospělé zákazníky.
Demografii jsme však dosud neupřesnili a může se výrazně lišit podle typu média.
Informace o konkrétních regionech zatím nemáme k dispozici.

Klíčové sdělení
Chceme zdůraznit kvalitu produktu, ale stále nevíme, jak přesně formulovat hlavní slogan.
… (Pracujeme na konceptu, který se musí schválit s naším interním oddělením.)

Rozpočet
Přesná výše dosud není stanovena, musíme počkat na vyjádření finančního oddělení.
Zatím pouze hrubý odhad, ale konkrétní čísla neznáme.

Časový harmonogram
Plánovaný start kampaně je „někdy v průběhu příštího čtvrtletí“.
Nemáme pevné termíny, protože dodání podkladů a kreativy zatím nebylo schváleno.

Víme, že je to velmi obecné a mnoho důležitých informací chybí. Jakmile se nám podaří získat upřesnění od vedení a dalších oddělení, rádi vám pošleme aktualizované podklady. Pokud potřebujete něco dalšího už nyní, dejte nám prosím vědět.

Děkujeme za trpělivost a omlouváme se za jakékoli komplikace.

S pozdravem,
Lukáš"""
```

```python
from langchain_core.messages import HumanMessage

msg = HumanMessage(content=f"Dostali jsme následující informace od klienta k media briefu. Můžeš prosím vypsat důležité chybící informace, které jsou nezbytné k optimalizaci mediálního plánování a zároveň posyktni argumentaci, proč jsou jednotlivé vhybějící proměné důležité a proč je pro všechny strany extrémně důležité, abychom tyto informace v briefu měli. Email:{email} ", name="Krystof")

message = [msg]

gpt4o_chat.invoke(message)
```
